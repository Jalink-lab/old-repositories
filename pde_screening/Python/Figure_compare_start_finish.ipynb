```python
import pandas as pd
from pathlib import Path
from AutomatedCellAnalysis.displaydata import displaydata
%matplotlib qt5
```

```python
base_path = Path('C:\\Users\\thinkBookUser\\Desktop\\Screening_Data_Analyzed_alldays')
my_csv1 = Path(base_path,'2019','12','05','caged','results','all_results.csv')
my_csv2 = Path(base_path,'2020','02','06','caged','results','all_results.csv')
out_pth = Path(base_path,'Figures')
title = 'Starting lifetime minus returning lifetime for caged cAMP'
```

```python
all_data1 = pd.read_csv(my_csv1)
all_data2 = pd.read_csv(my_csv2)
all_data = pd.concat([all_data1,all_data2])
errorfree_data = all_data[all_data['error'] == 0]
```

```python
# make start_finish
errorfree_data.loc[:,'finish tau(ns)'] = errorfree_data.loc[:,'start(ns)'].copy() + errorfree_data.loc[:,'range(ns)'].copy()
errorfree_data.loc[:,'return diff(ns)'] = errorfree_data.loc[:,'finish tau(ns)'].copy() - errorfree_data.loc[:,'start tau(ns)'].copy()
errorfree_data.head()
```

```python
fig, ax = displaydata(errorfree_data,'return diff(ns)')
ax.yaxis.set_label_text("")
fig.set_size_inches([17, 10])
ax.set_title(title)
```

```python
fig.savefig(Path(out_pth,title+'.png'),dpi=600)
```
