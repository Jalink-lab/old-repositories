```python
import pandas as pd
from pathlib import Path
from AutomatedCellAnalysis.displaydata import displaydata
%matplotlib qt5
```

```python
base_path = Path('C:\\Users\\thinkBookUser\\Desktop\\Screening_Data_Analyzed_alldays')
my_csv1 = Path(base_path,'2019','12','05','caged','results','all_results.csv')
my_csv2 = Path(base_path,'2020','02','06','caged','results','all_results.csv')
out_pth = Path(base_path,'Figures')

title = 'Supl.F. Caged cAMP starting lifetime'
```
