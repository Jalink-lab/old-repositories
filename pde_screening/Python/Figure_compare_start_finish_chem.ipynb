```python
import pandas as pd
from pathlib import Path
from AutomatedCellAnalysis.displaydata import displaydata
%matplotlib qt5
```

```python
base_path = Path('E:\\','Surfdrive','Screening_Data')
my_csv1 = Path(base_path,'2019','11','07','binned','all_results.csv')
my_csv2 = Path(base_path,'2020','02','06','Chemical','binned','all_results.csv')
my_csv3 = Path(base_path,'2020','01','07','binned','all_results.csv')
out_pth = Path('E:\\','Surfdrive','Screening_Data','out')
title = 'Lifetime before minus lifetime after for chemical stimulation starttau'
```

```python
all_data1 = pd.read_csv(my_csv1)
all_data2 = pd.read_csv(my_csv2)
all_data3 = pd.read_csv(my_csv3)
all_data1.condition = all_data1.condition.replace('Cilostamide_1uM', '1μM Cilostamide')
all_data1.condition = all_data1.condition.replace('IBMX_100_uM','100μM IBMX')
all_data2.condition = all_data2.condition.replace('IBMX_100_uM','100μM IBMX')
all_data2.condition = all_data2.condition.replace('Cilostamide_1uM','1μM Cilostamide')
all_data2.condition = all_data2.condition.replace('Double-KD_PDE3A&10A','Double-KD:\nPDE3A & PDE10A')
all_data2.condition = all_data2.condition.replace('100uM_IBMX_+_1uM_Cilostamide','1μM Cilostamide\n100μM IBMX')
all_data3.condition = all_data3.condition.replace('IBMX','100μM IBMX')
all_data3.condition = all_data3.condition.replace('ONLY_CELLS','Untransfected_cells')
all_data = pd.concat([all_data1,all_data2,all_data3])
all_data.condition = all_data.condition.replace('Untransfected_cells','Untransfected cells')
all_data.condition = all_data.condition.replace('RNAiMAX_reagent_only','RNAiMAX reagent only')
errorfree_data = all_data[all_data['error'] == 0]
```

```python
# make start_finish
errorfree_data.loc[:,'finish tau(ns)'] = errorfree_data.loc[:,'start(ns)'].copy() + errorfree_data.loc[:,'range(ns)'].copy()
errorfree_data.loc[:,'return diff(ns)'] = errorfree_data.loc[:,'finish tau(ns)'].copy() - errorfree_data.loc[:,'start tau(ns)'].copy()
# errorfree_data = errorfree_data[errorfree_data['finish tau(ns)']>1.8]
```

```python
# change the order
conditions = errorfree_data.condition.unique()
conditions_new = conditions.copy()
conditions_new[22] = conditions[25]
conditions_new[23:26] = conditions[22:25]
reordered_data = pd.DataFrame()
for condition in conditions_new:
    reordered_data = pd.concat([reordered_data, errorfree_data[errorfree_data['condition']==condition]])
reordered_data.condition.unique()
```

```python
fig, ax = displaydata(reordered_data,'start tau(ns)')
ax.yaxis.set_label_text("")
fig.set_size_inches([17, 10])
ax.set_title(title)
```

```python
fig.savefig(Path(out_pth,title+'.png'),dpi=600)
```
